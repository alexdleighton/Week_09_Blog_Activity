```python
import sqlite3
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("All four packages imported successfully.")
```

```python
conn = sqlite3.connect('week_09_data.db')

cursor = conn.cursor()
```

```python
pd.read_csv('sales.csv').to_sql('sales',conn, if_exists='replace',index=False)
```

```python
#Student_Practice_01

pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''',conn)
```

```python
pd.read_csv('hr_employees.csv').to_sql('employees',conn,if_exists='replace',index=False)
```

```python
#Student_Practice_02

pd.read_sql_query('''
SELECT name, department, salary
FROM employees
WHERE city = 'Chicago' AND years_experience > 10
''',conn)
```

```python
pd.read_csv('realestate_listings.csv').to_sql('listings',conn,if_exists='replace',index=False)
```

```python
#Student_Practice_03

pd.read_sql_query('''
SELECT *
FROM listings
ORDER BY sqft DESC
LIMIT 3
''',conn)
```

```python
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns',conn,if_exists='replace',index=False)
```

```python
#Student_Practice_04

pd.read_sql_query('''
SELECT campaign, SUM(impressions) AS total_impressions
FROM campaigns
GROUP BY campaign
''',conn)
```

```python
pd.read_csv('restaurant_orders.csv').to_sql('orders',conn,if_exists='replace',index=False)
```

```python
#Student_Practice_05

pd.read_sql_query('''
SELECT order_date, SUM(order_total) AS total_sales
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''',conn)
```

```python
pd.read_csv('clinic_visits.csv').to_sql('visits',conn,if_exists='replace',index=False)
```

```python
#Student_Practice_06

pd.read_sql_query('''
SELECT doctor, ROUND(AVG(patient_age),0) AS avg_age, MAX(wait_minutes) AS longest_wait
FROM visits
GROUP BY doctor
''',conn)
```

```python
pd.read_csv('bank_customers.csv').to_sql('customers',conn,if_exists='replace',index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions',conn,if_exists='replace',index=False)
```

```python
#Student_Practice_07

pd.read_sql_query('''
SELECT c.branch, SUM(t.amount) AS total_amount
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''',conn)
```
